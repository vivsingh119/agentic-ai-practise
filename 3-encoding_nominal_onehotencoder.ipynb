**Types of Categorical Data :**

**Nominal Categorical Data**
---
This is a type of categorical data
which is used to identify an object
uniquely without providing any additional information. We cannot rank them naturally.
Example:
     Names of Cities
     Name of a person.

**Ordinal Categorical Data**:
---
This is type of categorical data
which can be ranked naturally.
Example:
    Education Level, School Grades, Seniority Level
    Post Graduate School Graduate PhD
    
rank: PhD, Post Graduate, Graduate, School
