```python
import numpy as np
import pandas as pd;
from sklearn.preprocessing import OneHotEncoder
```

**Types of Categorical Data :**

**Nominal Categorical Data**
---
This is a type of categorical data
which is used to identify an object
uniquely without providing any additional information. We cannot rank them naturally.
Example:
     Names of Cities
     Name of a person.

**Ordinal Categorical Data**:
---
This is type of categorical data
which can be ranked naturally.
Example:
    Education Level, School Grades, Seniority Level
    Post Graduate School Graduate PhD
    
rank: PhD, Post Graduate, Graduate, School

```python
df_township=pd.read_csv("https://raw.githubusercontent.com/vivsingh119/agentic-ai-practise/refs/heads/master/township.csv")
df_township
```

```python
df_township.columns
```

```python
df_township.dtypes
```

```python
df_township['town'].unique()
```

```python
df_township['town'].isna().sum()
```

```python
ohe=OneHotEncoder()
```

```python
ohe.fit(df_township[['town']])
```

```python
ohe.transform(df_township[['town']])
```

```python
encoded_df=ohe.transform(df_township[['town']]).toarray()
encoded_df
```

```python
ohe.categories_
```

```python
df_new=pd.DataFrame(encoded_df,columns=ohe.categories_,dtype=int)
df_new
```

```python
pd.concat([df_township,df_new],axis=1)
```
